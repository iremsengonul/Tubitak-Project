```python
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import pulp
from IPython.display import display

# =====================================================================
# MODEL PARAMETRELERİ
# =====================================================================
SPEED_KMH = 50.0
# Ürünlerin depodan talep noktasına ulaşabileceği en uzun süre
MAX_TRAVEL_HOURS = 3
# İlçe nüfusunun ilaç ihtiyacı olan kısmının oranı
NEED_RATE = 0.09
# Bir depo için izin verilen en yüksek kapasite
BASE_DEPOT_CAPACITY = 600_000
# Tıbbi ilaç dolabı varsayımları
MEDICINE_CABINET_LITRES = 400
CABINET_USABLE_FILL_RATE = 0.70
AVG_MEDICINE_PACKAGE_LITRES = 0.40
MEDICINES_PER_CABINET = math.floor(
    MEDICINE_CABINET_LITRES
    * CABINET_USABLE_FILL_RATE
    / AVG_MEDICINE_PACKAGE_LITRES
)
# Bir tıbbi ilaç dolabının günlük enerji tüketimi
ENERGY_PER_CABINET_KWH_DAY = 5.28
# Hastane taban alanının soğuk depolamaya ayrıldığı varsayılan oran
COLD_STORAGE_AREA_SHARE = 0.0085
# Bir dolap için gerekli yaklaşık alan
SPACE_PER_CABINET_M2 = 1.50
# Tüm açık depoların toplam günlük enerji sınırı
MAX_TOTAL_ENERGY_KWH_DAY = 3000
USE_OSRM = True
ROAD_FACTOR_FALLBACK = 1.20
SOLVER_TIME_LIMIT_SECONDS = 300

# =====================================================================
# 2025 İZMİR İLÇE NÜFUSLARI
# =====================================================================
DEMAND_POINTS = {
    "I01_ALIAGA": {
        "name": "ALİAĞA",
        "lat": 38.798476,
        "lon": 27.009075,
        "population": 110_892,
    },
    "I02_BALCOVA": {
        "name": "BALÇOVA",
        "lat": 38.386271,
        "lon": 27.057597,
        "population": 75_743,
    },
    "I03_BAYINDIR": {
        "name": "BAYINDIR",
        "lat": 38.223108,
        "lon": 27.628759,
        "population": 39_971,
    },
    "I04_BAYRAKLI": {
        "name": "BAYRAKLI",
        "lat": 38.470998,
        "lon": 27.158308,
        "population": 296_481,
    },
    "I05_BERGAMA": {
        "name": "BERGAMA",
        "lat": 39.161089,
        "lon": 27.214566,
        "population": 107_549,
    },
    "I06_BORNOVA": {
        "name": "BORNOVA",
        "lat": 38.459001,
        "lon": 27.229283,
        "population": 452_479,
    },
    "I07_BUCA": {
        "name": "BUCA",
        "lat": 38.376772,
        "lon": 27.178878,
        "population": 520_941,
    },
    "I08_CESME": {
        "name": "ÇEŞME",
        "lat": 38.317326,
        "lon": 26.344325,
        "population": 51_902,
    },
    "I09_CIGLI": {
        "name": "ÇİĞLİ",
        "lat": 38.504458,
        "lon": 27.054772,
        "population": 216_430,
    },
    "I10_DIKILI": {
        "name": "DİKİLİ",
        "lat": 39.075001,
        "lon": 26.900312,
        "population": 49_138,
    },
    "I11_FOCA": {
        "name": "FOÇA",
        "lat": 38.691060,
        "lon": 26.824303,
        "population": 37_711,
    },
    "I12_GAZIEMIR": {
        "name": "GAZİEMİR",
        "lat": 38.320504,
        "lon": 27.140987,
        "population": 137_220,
    },
    "I13_KARABAGLAR": {
        "name": "KARABAĞLAR",
        "lat": 38.384666,
        "lon": 27.104072,
        "population": 467_572,
    },
    "I14_KARABURUN": {
        "name": "KARABURUN",
        "lat": 38.592233,
        "lon": 26.492988,
        "population": 13_725,
    },
    "I15_KARSIYAKA": {
        "name": "KARŞIYAKA",
        "lat": 38.470547,
        "lon": 27.102063,
        "population": 339_624,
    },
    "I16_KEMALPASA": {
        "name": "KEMALPAŞA",
        "lat": 38.426377,
        "lon": 27.485666,
        "population": 122_840,
    },
    "I17_KINIK": {
        "name": "KINIK",
        "lat": 39.064684,
        "lon": 27.380008,
        "population": 28_604,
    },
    "I18_KIRAZ": {
        "name": "KİRAZ",
        "lat": 38.214458,
        "lon": 28.288491,
        "population": 43_340,
    },
    "I19_KONAK": {
        "name": "KONAK",
        "lat": 38.417914,
        "lon": 27.142638,
        "population": 317_399,
    },
    "I20_MENDERES": {
        "name": "MENDERES",
        "lat": 38.180933,
        "lon": 27.125942,
        "population": 112_853,
    },
    "I21_MENEMEN": {
        "name": "MENEMEN",
        "lat": 38.607873,
        "lon": 27.067971,
        "population": 220_510,
    },
    "I22_ODEMIS": {
        "name": "ÖDEMİŞ",
        "lat": 38.211444,
        "lon": 27.993760,
        "population": 132_548,
    },
    "I23_SEFERIHISAR": {
        "name": "SEFERİHİSAR",
        "lat": 38.178635,
        "lon": 26.869247,
        "population": 63_712,
    },
    "I24_SELCUK": {
        "name": "SELÇUK",
        "lat": 37.945601,
        "lon": 27.374520,
        "population": 38_578,
    },
    "I25_TIRE": {
        "name": "TİRE",
        "lat": 38.072962,
        "lon": 27.727698,
        "population": 87_571,
    },
    "I26_TORBALI": {
        "name": "TORBALI",
        "lat": 38.188111,
        "lon": 27.361339,
        "population": 224_133,
    },
    "I27_URLA": {
        "name": "URLA",
        "lat": 38.336587,
        "lon": 26.736737,
        "population": 82_021,
    },
}
# Talep, ilçe nüfusunun ilaç ihtiyacı oranı kadar alınır
for point in DEMAND_POINTS.values():
    point["demand"] = math.ceil(
        point["population"] * NEED_RATE
    )

# =====================================================================
# HASTANE / DEPO ADAYLARI
# =====================================================================
#
# ground_area_m2 = hastanenin yaklaşık bina zemin oturum alanı.
# Kat sayısı kullanılmıyor.
#
DEPOT_CANDIDATES = {
    "D01_ALIAGA": {
        "name": "Aliağa Devlet Hastanesi",
        "lat": 38.787045,
        "lon": 26.974983,
        "ground_area_m2": 2_470,
    },
    "D02_BALCOVA": {
        "name": "Dokuz Eylül Üniversitesi Hastanesi",
        "lat": 38.394177,
        "lon": 27.033105,
        "ground_area_m2": 31_157,
    },
    "D03_BAYINDIR": {
        "name": "Bayındır Devlet Hastanesi",
        "lat": 38.221535,
        "lon": 27.649601,
        "ground_area_m2": 1_538,
    },
    "D04_BAYRAKLI": {
        "name": "İzmir Şehir Hastanesi",
        "lat": 38.484367,
        "lon": 27.177654,
        "ground_area_m2": 47_238,
    },
    "D05_BERGAMA": {
        "name": "Bergama Devlet Hastanesi",
        "lat": 39.118364,
        "lon": 27.197994,
        "ground_area_m2": 6_211,
    },
    "D06_BORNOVA": {
        "name": "Ege Üniversitesi Hastanesi",
        "lat": 38.456052,
        "lon": 27.209807,
        "ground_area_m2": 88_627,
    },
    "D07_BUCA": {
        "name": "Buca Seyfi Demirsoy Hastanesi",
        "lat": 38.384522,
        "lon": 27.163315,
        "ground_area_m2": 4_079,
    },
    "D08_CESME": {
        "name": "Çeşme Devlet Hastanesi",
        "lat": 38.316859,
        "lon": 26.325647,
        "ground_area_m2": 4_804,
    },
    "D09_CIGLI": {
        "name": "Çiğli Eğitim ve Araştırma Hastanesi",
        "lat": 38.499592,
        "lon": 27.049329,
        "ground_area_m2": 18_449,
    },
    "D10_DIKILI": {
        "name": "Dikili Devlet Hastanesi",
        "lat": 39.059778,
        "lon": 26.891532,
        "ground_area_m2": 4_997,
    },
    "D11_FOCA": {
        "name": "Foça Devlet Hastanesi",
        "lat": 38.658172,
        "lon": 26.761039,
        "ground_area_m2": 3_705,
    },
    "D12_GAZIEMIR": {
        "name": "Gaziemir Devlet Hastanesi",
        "lat": 38.311027,
        "lon": 27.133142,
        "ground_area_m2": 2_729,
    },
    "D13_KARABAGLAR": {
        "name": "Atatürk Eğitim ve Araştırma Hastanesi",
        "lat": 38.397750,
        "lon": 27.103803,
        "ground_area_m2": 17_953,
    },
    "D14_KARABURUN": {
        "name": "Karaburun İlçe Devlet Hastanesi",
        "lat": 38.635182,
        "lon": 26.522338,
        "ground_area_m2": 1_050,
    },
    "D15_KARSIYAKA": {
        "name": "Başkent Üniversitesi Zübeyde Hanım Hastanesi",
        "lat": 38.471847,
        "lon": 27.086234,
        "ground_area_m2": 266,
    },
    "D16_KEMALPASA": {
        "name": "Kemalpaşa Devlet Hastanesi",
        "lat": 38.436253,
        "lon": 27.409816,
        "ground_area_m2": 1_794,
    },
    "D17_KINIK": {
        "name": "Kınık Devlet Hastanesi",
        "lat": 39.089646,
        "lon": 27.362965,
        "ground_area_m2": 2_244,
    },
    "D18_KIRAZ": {
        "name": "Kiraz Devlet Hastanesi",
        "lat": 38.230386,
        "lon": 28.218448,
        "ground_area_m2": 2_554,
    },
    "D19_KONAK": {
        "name": "Eşrefpaşa Hastanesi",
        "lat": 38.423557,
        "lon": 27.160669,
        "ground_area_m2": 5_096,
    },
    "D20_MENDERES": {
        "name": "Menderes Devlet Hastanesi",
        "lat": 38.238470,
        "lon": 27.136158,
        "ground_area_m2": 4_833,
    },
    "D21_MENEMEN": {
        "name": "Menemen Devlet Hastanesi",
        "lat": 38.606998,
        "lon": 27.088010,
        "ground_area_m2": 10_459,
    },
    "D22_ODEMIS": {
        "name": "Ödemiş Devlet Hastanesi",
        "lat": 38.230523,
        "lon": 27.988069,
        "ground_area_m2": 7_276,
    },
    "D23_SEFERIHISAR": {
        "name": "Seferihisar Necat Hepkon Devlet Hastanesi",
        "lat": 38.197890,
        "lon": 26.845544,
        "ground_area_m2": 1_812,
    },
    "D24_SELCUK": {
        "name": "Selçuk Devlet Hastanesi",
        "lat": 37.947827,
        "lon": 27.367257,
        "ground_area_m2": 2_032,
    },
    "D25_TIRE": {
        "name": "Tire Devlet Hastanesi",
        "lat": 38.100611,
        "lon": 27.721758,
        "ground_area_m2": 7_535,
    },
    "D26_TORBALI": {
        "name": "Torbalı Devlet Hastanesi",
        "lat": 38.175630,
        "lon": 27.358065,
        "ground_area_m2": 7_605,
    },
    "D27_URLA": {
        "name": "Urla Devlet Hastanesi",
        "lat": 38.330189,
        "lon": 26.745952,
        "ground_area_m2": 6_875,
    },
}

# =====================================================================
# HASTANE ALANI -> DOLAP SAYISI -> KAPASİTE -> ENERJİ
# =====================================================================
for depot in DEPOT_CANDIDATES.values():
    # Yaklaşık kullanılabilir soğuk depo alanı
    cold_storage_area = (
        depot["ground_area_m2"]
        * COLD_STORAGE_AREA_SHARE
    )
    # Hastaneye yaklaşık kaç ilaç dolabı sığabilir?
    depot["cabinet_count"] = max(
        1,
        math.floor(
            cold_storage_area
            / SPACE_PER_CABINET_M2
        ),
    )
    # Dolap sayısına göre kapasite
    calculated_capacity = (
        depot["cabinet_count"]
        * MEDICINES_PER_CABINET
    )
    # Hesaplanan kapasiteyi depo başına kapasite üst sınırıyla sınırlar.
    depot["capacity"] = min(
        BASE_DEPOT_CAPACITY,
        calculated_capacity,
    )
    # Dolap sayısına göre enerji
    depot["energy_kwh_day"] = (
        depot["cabinet_count"]
        * ENERGY_PER_CABINET_KWH_DAY
    )

# =====================================================================
# VERİ SETİ ÖZETİ
# =====================================================================

print("=" * 70)
print("VERİ SETİ ÖZETİ")
print("=" * 70)
print(
    "2025 İzmir toplam nüfusu:",
    f"{sum(p['population'] for p in DEMAND_POINTS.values()):,}"
)
print(
    "Toplam ilaç ihtiyacı:",
    f"{sum(p['demand'] for p in DEMAND_POINTS.values()):,}"
)
print(
    "Bir dolabın kapasitesi:",
    f"{MEDICINES_PER_CABINET:,}",
    "paket eşdeğeri"
)
print(
    "Bir dolabın enerjisi:",
    f"{ENERGY_PER_CABINET_KWH_DAY:.2f}",
    "kWh/gün"
)
print(
    "\nToplam enerji sınırı:",
    f"{MAX_TOTAL_ENERGY_KWH_DAY:.2f} kWh/gün"
)

# =====================================================================
# MESAFE
# =====================================================================
def haversine_km(lat1, lon1, lat2, lon2):
    R = 6371.0088
    p1 = math.radians(lat1)
    p2 = math.radians(lat2)
    dlat = math.radians(lat2 - lat1)
    dlon = math.radians(lon2 - lon1)
    a = (
        math.sin(dlat / 2) ** 2
        + math.cos(p1)
        * math.cos(p2)
        * math.sin(dlon / 2) ** 2
    )
    return 2 * R * math.asin(math.sqrt(a))
def build_distance_matrix():
    """
    İnternet varsa OSRM yol mesafesi.
    OSRM çalışmazsa Haversine mesafesi yol katsayısıyla çarpılır.
    """
    depot_ids = list(DEPOT_CANDIDATES)
    demand_ids = list(DEMAND_POINTS)
    if USE_OSRM:
        try:
            import requests
            ordered = (
                [DEPOT_CANDIDATES[j] for j in depot_ids]
                + [DEMAND_POINTS[i] for i in demand_ids]
            )
            coords = ";".join(
                f"{p['lon']},{p['lat']}"
                for p in ordered
            )
            sources = ";".join(
                str(k)
                for k in range(len(depot_ids))
            )
            destinations = ";".join(
                str(k)
                for k in range(
                    len(depot_ids),
                    len(ordered)
                )
            )
            url = (
                "https://router.project-osrm.org/"
                f"table/v1/driving/{coords}"
            )
            response = requests.get(
                url,
                params={
                    "annotations": "distance",
                    "sources": sources,
                    "destinations": destinations,
                },
                timeout=60,
            )
            response.raise_for_status()
            data = response.json()
            distance = {}
            for j_index, j in enumerate(depot_ids):
                for i_index, i in enumerate(demand_ids):
                    meters = (
                        data["distances"][j_index][i_index]
                    )
                    if meters is None:
                        raise RuntimeError(
                            "OSRM bir rota için mesafe döndürmedi."
                        )
                    distance[(i, j)] = round(
                        meters / 1000,
                        3
                    )
            return distance, "OSRM yol mesafesi"
        except Exception as error:
            print(
                "OSRM kullanılamadı:",
                error
            )
            print(
                "Haversine x 1.20 yaklaşımı kullanılacak."
            )
    distance = {}
    for i, demand in DEMAND_POINTS.items():
        for j, depot in DEPOT_CANDIDATES.items():
            km = haversine_km(
                demand["lat"],
                demand["lon"],
                depot["lat"],
                depot["lon"],
            )
            distance[(i, j)] = round(
                max(
                    0.1,
                    km * ROAD_FACTOR_FALLBACK
                ),
                3
            )
    return distance, "Haversine x 1.20"

# =====================================================================
# GUROBI
# =====================================================================
def solve_with_gurobi(model):
    solver = pulp.GUROBI(
        msg=True,
        timeLimit=SOLVER_TIME_LIMIT_SECONDS,
        gapRel=0.0,
    )
    if not solver.available():
        raise RuntimeError(
            "Gurobi bulunamadı veya lisans etkin değil."
        )
    result = model.solve(solver)
    # PuLP 4 ve sonrası çözüm bilgisini nesne olarak, eski sürümler durum kodu olarak döndürür.
    if hasattr(result, "status_str"):
        if result.status_str == "Optimal":
            return "Optimal"
        return "Feasible" if result.has_solution else result.status_str
    return pulp.LpStatus[result]

# =====================================================================
# GÖRSELLEŞTİRME
# =====================================================================
def local_xy(lat, lon, lat0, lon0):
    x = (
        (lon - lon0)
        * 111.320
        * math.cos(math.radians(lat0))
    )
    y = (
        (lat - lat0)
        * 110.574
    )
    return x, y
def show_solution(distance, open_depots, assignments, title):
    # Önceki grafiklerden kalan figürlerin tekrar görüntülenmesini önler.
    plt.close("all")
    all_points = list(DEMAND_POINTS.values()) + list(DEPOT_CANDIDATES.values())
    lat0 = np.mean([p["lat"] for p in all_points])
    lon0 = np.mean([p["lon"] for p in all_points])
    # Açık depolar için birbirinden belirgin biçimde ayrışan renkler.
    distinct_colors = [
        "#E6194B", "#0072B2", "#3CB44B", "#FF8C00", "#911EB4", "#00BCD4",
        "#F032E6", "#8B4513", "#C8B400", "#000080", "#7FD000", "#008080",
    ]
    # Açılmayan depo adayları için sabit renk.
    CLOSED_DEPOT_COLOR = "#D0D0D0"
    # Coğrafi koordinatları kilometre cinsinden yerel düzlem koordinatlarına dönüştürür.
    depot_xy = {}
    for j, depot in DEPOT_CANDIDATES.items():
        depot_xy[j] = local_xy(depot["lat"], depot["lon"], lat0, lon0)
    # Birbirine yakın açık depolara birbirine en az benzeyen renkleri atar.
    def color_gap(c1, c2):
        r1, g1, b1 = (int(c1[k:k + 2], 16) for k in (1, 3, 5))
        r2, g2, b2 = (int(c2[k:k + 2], 16) for k in (1, 3, 5))
        rm = (r1 + r2) / 2
        return math.sqrt(
            (2 + rm / 256) * (r1 - r2) ** 2
            + 4 * (g1 - g2) ** 2
            + (2 + (255 - rm) / 256) * (b1 - b2) ** 2
        )
    def geo_gap(j, k):
        return math.hypot(depot_xy[j][0] - depot_xy[k][0], depot_xy[j][1] - depot_xy[k][1])
    depot_colors = {j: CLOSED_DEPOT_COLOR for j in DEPOT_CANDIDATES}
    crowd_order = sorted(
        open_depots,
        key=lambda j: sum(1 / (1 + geo_gap(j, k)) for k in open_depots if k != j),
        reverse=True,
    )
    colored = []
    for j in crowd_order:
        used = {depot_colors[k] for k in colored}
        candidates = [c for c in distinct_colors if c not in used] or distinct_colors
        depot_colors[j] = max(
            candidates,
            key=lambda c: min(
                (color_gap(c, depot_colors[k]) + 15 * geo_gap(j, k) for k in colored),
                default=0,
            ),
        )
        colored.append(j)
    # Talep noktalarını yalnızca görsel çakışmaları azaltacak kadar kaydırır; model verilerini değiştirmez.
    demand_xy = {}
    MIN_VISUAL_DISTANCE = 4.5
    VISUAL_SHIFT = 3.2
    placed_points = list(depot_xy.values())
    for index, (i, point) in enumerate(DEMAND_POINTS.items()):
        xi, yi = local_xy(point["lat"], point["lon"], lat0, lon0)
        plot_x, plot_y = xi, yi
        attempt = 0
        while any(
            math.hypot(plot_x - px, plot_y - py) < MIN_VISUAL_DISTANCE
            for px, py in placed_points
        ):
            angle = math.radians((index * 53 + attempt * 71) % 360)
            radius = VISUAL_SHIFT + attempt * 1.0
            plot_x = xi + radius * math.cos(angle)
            plot_y = yi + radius * math.sin(angle)
            attempt += 1
            if attempt > 16:
                break
        demand_xy[i] = (plot_x, plot_y)
        placed_points.append((plot_x, plot_y))
    # ------------------------- 1. HARİTA GRAFİĞİ -------------------------
    fig, ax = plt.subplots(figsize=(24, 18))
    depot_offsets = [
        (7, 8), (-7, 8), (7, -10), (-7, -10),
        (11, 2), (-11, 2), (10, 10), (-10, -10)
    ]
    # Açık depoları kendi renkleriyle, açılmayanları sabit renkle kare işaretçiyle gösterir.
    for index, (j, depot) in enumerate(DEPOT_CANDIDATES.items()):
        xj, yj = depot_xy[j]
        is_open = j in open_depots
        ax.scatter(
            xj, yj,
            marker="s",
            s=650 if is_open else 320,
            c=depot_colors[j],
            edgecolors="none",
            alpha=1.0,
            zorder=7 if is_open else 4
        )
        dx, dy = depot_offsets[index % len(depot_offsets)]
        ax.annotate(
            j,
            (xj, yj),
            xytext=(dx, dy),
            textcoords="offset points",
            fontsize=7.5,
            ha="center",
            va="center",
            zorder=12,
            bbox=dict(boxstyle="round,pad=0.10", fc="white", ec="none", alpha=0.70)
        )
    label_offsets = [
        (12, 12), (-12, 12), (12, -14), (-12, -14),
        (18, 2), (-18, 2), (17, 14), (-17, 14),
        (17, -14), (-17, -14), (3, 18), (3, -18)
    ]
    # Tüm talep noktalarını aynı renk ve boyuttaki dairelerle gösterir.
    for index, (i, point) in enumerate(DEMAND_POINTS.items()):
        xi, yi = demand_xy[i]
        size = 200
        ax.scatter(
            xi, yi,
            marker="o",
            s=size,
            c="#5F5F5F",
            edgecolors="white",
            linewidths=1.8,
            alpha=0.90,
            zorder=8
        )
        dx, dy = label_offsets[index % len(label_offsets)]
        ax.annotate(
            point["name"],
            (xi, yi),
            xytext=(dx, dy),
            textcoords="offset points",
            fontsize=8.0,
            ha="center",
            va="center",
            zorder=13,
            bbox=dict(boxstyle="round,pad=0.10", fc="white", ec="none", alpha=0.72)
        )
        # Talep noktasını hizmet aldığı deponun rengiyle bağlayarak atamayı görünür kılar.
        j = assignments[i]
        xj, yj = depot_xy[j]
        ax.plot(
            [xj, xi],
            [yj, yi],
            color=depot_colors[j],
            linewidth=4.5,
            alpha=0.9,
            solid_capstyle="round",
            zorder=3
        )
    ax.set_title(title, fontsize=15, pad=20)
    ax.set_xlabel("Doğu-Batı uzaklığı (km)")
    ax.set_ylabel("Kuzey-Güney uzaklığı (km)")
    # Eksen değerlerinden çıkan eşit aralıklı kare ızgarayı belirgin ama açık tonda çizer.
    from matplotlib.ticker import MultipleLocator
    ax.xaxis.set_major_locator(MultipleLocator(20))
    ax.yaxis.set_major_locator(MultipleLocator(20))
    ax.grid(True, which="major", linestyle="-", linewidth=0.9, color="#8C8C8C", alpha=0.55)
    ax.set_axisbelow(True)
    ax.set_aspect("equal", adjustable="datalim")
    ax.margins(x=0.18, y=0.18)
    # Harita sembollerinin anlamını depo renklerinden bağımsız olarak açıklar.
    from matplotlib.lines import Line2D
    legend_handles = [
        Line2D([0], [0], marker="o", color="none", markerfacecolor="#5F5F5F",
               markeredgecolor="white", markersize=10, label="Talep noktası"),
        Line2D([0], [0], marker="s", color="none", markerfacecolor=depot_colors[open_depots[0]],
               markeredgecolor="none", markersize=13, label="Açık depo"),
        Line2D([0], [0], marker="s", color="none", markerfacecolor=CLOSED_DEPOT_COLOR,
               markeredgecolor="none", markersize=11, label="Açılmayan depo adayı"),
        Line2D([0], [0], color="black", linewidth=4.5, label="Depo-talep ataması")
    ]
    ax.legend(handles=legend_handles, loc="upper left", bbox_to_anchor=(1.01, 1))
    plt.tight_layout()
    plt.show()
    plt.close(fig)
    # ---------------------- 2. YÜK-KAPASİTE GRAFİĞİ ----------------------
    loads = [
        sum(DEMAND_POINTS[i]["demand"] for i in DEMAND_POINTS if assignments[i] == j)
        for j in open_depots
    ]
    capacities = [DEPOT_CANDIDATES[j]["capacity"] for j in open_depots]
    x_pos = np.arange(len(open_depots))
    width = 0.36
    fig, ax = plt.subplots(figsize=(13, 7))
    ax.bar(x_pos - width / 2, loads, width, label="Atanan talep")
    ax.bar(x_pos + width / 2, capacities, width, label="Depo kapasitesi")
    # Her deponun kapasite kullanım oranını iki sütunun ortasına yazar.
    for k, (load, cap) in enumerate(zip(loads, capacities)):
        ax.text(
            x_pos[k],
            max(load, cap) * 1.01,
            f"%{100 * load / cap:.1f}".replace(".", ","),
            ha="center",
            va="bottom",
            fontsize=9,
            fontweight="bold",
        )
    ax.set_ylim(0, max(capacities) * 1.08)
    ax.set_xticks(x_pos, open_depots, rotation=45, ha="right")
    ax.set_ylabel("Talep / kapasite")
    ax.set_title("Açık Depoların Yük ve Kapasite Karşılaştırması")
    ax.grid(True, axis="y", linestyle="--", alpha=0.4)
    ax.legend()
    plt.tight_layout()
    plt.show()
    plt.close(fig)
    # ------------------------- 3. DAİRE GRAFİĞİ -------------------------
    pie_loads = [
        sum(DEMAND_POINTS[i]["demand"] for i in DEMAND_POINTS if assignments[i] == j)
        for j in open_depots
    ]
    pie_labels = list(open_depots)
    pie_colors = [depot_colors[j] for j in open_depots]
    fig, ax = plt.subplots(figsize=(11, 9))
    wedges, texts, autotexts = ax.pie(
        pie_loads,
        labels=pie_labels,
        colors=pie_colors,
        autopct="%1.1f%%",
        startangle=90,
        pctdistance=0.75,
        labeldistance=1.05,
        wedgeprops=dict(edgecolor="white", linewidth=1.5)
    )
    ax.set_title("Açık Depoların Toplam Talep İçindeki Payı", pad=20)
    legend_labels = [f"{j} - {DEPOT_CANDIDATES[j]['name']}" for j in open_depots]
    ax.legend(
        wedges,
        legend_labels,
        title="Açık Depolar",
        loc="center left",
        bbox_to_anchor=(1.0, 0.5)
    )
    plt.tight_layout()
    plt.show()
    plt.close(fig)

# =====================================================================
# MESAFE VE ULAŞIM SÜRESİ MATRİSLERİ
# =====================================================================
distance, distance_source = build_distance_matrix()
travel_time = {
    (i, j): distance[(i, j)] / SPEED_KMH
    for i in DEMAND_POINTS
    for j in DEPOT_CANDIDATES
}

# =====================================================================
# KISITLAR
# =====================================================================
def add_constraints(
    model,
    x,
    y,
):
    # ---------------------------------------------------------
    # (1) Her talep noktası bir depodan hizmet almalı
    # ---------------------------------------------------------
    for i in DEMAND_POINTS:
        model += (
            pulp.lpSum(
                y[(i, j)]
                for j in DEPOT_CANDIDATES
            )
            == 1
        )
    # ---------------------------------------------------------
    # (2) Depo açılmadan hizmet veremez
    # ---------------------------------------------------------
    for i in DEMAND_POINTS:
        for j in DEPOT_CANDIDATES:
            model += (
                y[(i, j)]
                <= x[j]
            )
    # ---------------------------------------------------------
    # (3) Depo kapasitesi aşılmamalı
    # ---------------------------------------------------------
    for j in DEPOT_CANDIDATES:
        model += (
            pulp.lpSum(
                DEMAND_POINTS[i]["demand"]
                * y[(i, j)]
                for i in DEMAND_POINTS
            )
            <=
            DEPOT_CANDIDATES[j]["capacity"]
            * x[j]
        )
    # ---------------------------------------------------------
    # (4) Ürün bozulmadan ulaştırılmalı
    # ---------------------------------------------------------
    for i in DEMAND_POINTS:
        for j in DEPOT_CANDIDATES:
            #print("Hastane Kapasiteleri********")
            #print(DEPOT_CANDIDATES[j]["capacity"]* x[j])
            # print("Hastane Enerjileri *******")
            #print(DEPOT_CANDIDATES[j]["energy_kwh_day"] * x[j])
            model += (
                travel_time[(i, j)]
                * y[(i, j)]
                <= MAX_TRAVEL_HOURS
            )
    # ---------------------------------------------------------
    # (5) Açık depoların toplam günlük enerji tüketimi sınırı aşmamalı
    # ---------------------------------------------------------
    model += (
        pulp.lpSum(
            DEPOT_CANDIDATES[j]["energy_kwh_day"] * x[j]
            for j in DEPOT_CANDIDATES
        )
        <= MAX_TOTAL_ENERGY_KWH_DAY
    )
    

# =====================================================================
# DEĞİŞKENLER
# =====================================================================
def new_variable(model, name, lowBound=None, upBound=None, cat="Continuous"):
    # PuLP 4 ve sonrası değişkeni problem üzerinden, eski sürümler doğrudan oluşturur.
    if hasattr(model, "add_variable"):
        return model.add_variable(name, lowBound, upBound, cat)
    return pulp.LpVariable(name, lowBound, upBound, cat)
def create_binary_variables(model, prefix):
    x = {
        j: new_variable(model, f"{prefix}_depo_acik_{j}", cat="Binary")
        for j in DEPOT_CANDIDATES
    }
    y = {
        (i, j): new_variable(model, f"{prefix}_atama_{i}_{j}", cat="Binary")
        for i in DEMAND_POINTS
        for j in DEPOT_CANDIDATES
    }
    return x, y

# =====================================================================
# AŞAMA 1
#
# En az kaç depo gerektiğini bul
# =====================================================================
stage1 = pulp.LpProblem(
    "Izmir_Minimum_Depo_Sayisi",
    pulp.LpMinimize,
)
x1, y1 = create_binary_variables(
    stage1,
    "asama1"
)
stage1 += pulp.lpSum(
    x1[j]
    for j in DEPOT_CANDIDATES
)
add_constraints(
    stage1,
    x1,
    y1,
)
status1 = solve_with_gurobi(
    stage1
)
print(
    "Aşama 1 durumu:",
    status1
)
if status1 not in {
    "Optimal",
    "Feasible",
}:
    raise RuntimeError(
        "Minimum depo sayısı bulunamadı."
    )
p_star = int(
    round(
        sum(
            pulp.value(x1[j])
            for j in DEPOT_CANDIDATES
        )
    )
)
print(
    "Minimum gerekli depo sayısı:",
    p_star
)

# =====================================================================
# AŞAMA 2 - Minimum depo sayısını koruyarak minimax değeri bul
# =====================================================================
stage2=pulp.LpProblem("Izmir_Optimal_Depo_Sayili_Minimax",pulp.LpMinimize)
x2,y2=create_binary_variables(stage2,"asama2")
z=new_variable(stage2,"maksimum_agirlikli_mesafe",lowBound=0)
stage2+=z
add_constraints(stage2,x2,y2)
stage2+=pulp.lpSum(x2[j] for j in DEPOT_CANDIDATES)==p_star
# Her talep noktasının ağırlıklı mesafesini z üst sınırının altında tutar.
for i in DEMAND_POINTS: stage2+=z>=pulp.lpSum(DEMAND_POINTS[i]["demand"]*distance[(i,j)]*y2[(i,j)] for j in DEPOT_CANDIDATES)
status2=solve_with_gurobi(stage2)
print("Aşama 2 durumu:",status2)
print("Mesafe kaynağı:",distance_source)
if status2 not in {"Optimal","Feasible"}: raise RuntimeError("Minimax çözümü bulunamadı.")
z_star=pulp.value(z)
print(f"Optimum minimax değeri: {z_star:,.2f} ihtiyaç-km")

# =====================================================================
# AŞAMA 3 - Minimax optimumunu koruyarak toplam mesafeyi minimize et
# =====================================================================
stage3=pulp.LpProblem("Izmir_Minimax_Korumali_Minimum_Toplam_Mesafe",pulp.LpMinimize)
x3,y3=create_binary_variables(stage3,"asama3")
# Amaç, bütün talep-depo atamalarının talep ağırlıklı toplam yol mesafesidir.
stage3+=pulp.lpSum(DEMAND_POINTS[i]["demand"]*distance[(i,j)]*y3[(i,j)] for i in DEMAND_POINTS for j in DEPOT_CANDIDATES)
add_constraints(stage3,x3,y3)
stage3+=pulp.lpSum(x3[j] for j in DEPOT_CANDIDATES)==p_star
# Aşama 2'de bulunan optimum minimax seviyesinin kötüleşmesini engeller.
for i in DEMAND_POINTS: stage3+=pulp.lpSum(DEMAND_POINTS[i]["demand"]*distance[(i,j)]*y3[(i,j)] for j in DEPOT_CANDIDATES)<=z_star+1e-6
status3=solve_with_gurobi(stage3)
print("Aşama 3 durumu:",status3)
if status3 not in {"Optimal","Feasible"}: raise RuntimeError("Minimax optimumunu koruyan minimum toplam mesafe çözümü bulunamadı.")

# =====================================================================
# SONUÇLAR
# =====================================================================
open_depots = [
    j
    for j in DEPOT_CANDIDATES
    if (
        pulp.value(x3[j]) is not None
        and pulp.value(x3[j]) > 0.5
    )
]
assignments = {
    i: next(
        j
        for j in DEPOT_CANDIDATES
        if (
            pulp.value(y3[(i, j)]) is not None
            and pulp.value(y3[(i, j)]) > 0.5
        )
    )
    for i in DEMAND_POINTS
}
rows = []
for i, j in assignments.items():
    rows.append({
        "talep_noktası":
            DEMAND_POINTS[i]["name"],
        "2025_nüfus":
            DEMAND_POINTS[i]["population"],
        "talep":
            DEMAND_POINTS[i]["demand"],
        "atanan_depo":
            DEPOT_CANDIDATES[j]["name"],
        "hastane_kapasitesi":
            DEPOT_CANDIDATES[j]["capacity"],
        "mesafe_km":
            distance[(i, j)],
        "süre_dakika":
            distance[(i, j)]
            / SPEED_KMH
            * 60,
    })
result = pd.DataFrame(
    rows
)
display(
    result
)
print(f"Toplam talep: {result['talep'].sum():,} paket")

# =====================================================================
# HASTANE KAPASİTE VE ENERJİ TABLOSU
# =====================================================================
# Açılan ve açılmayan bütün hastanelerin kapasitesini ve günlük enerji tüketimini gösterir.
hospital_rows = []
for j, depot in DEPOT_CANDIDATES.items():
    hospital_rows.append({
        "hastane":
            depot["name"],
        "hastane_kapasitesi":
            depot["capacity"],
        "enerji_tüketimi_kWh_gün":
            round(depot["energy_kwh_day"], 2),
        "durum":
            "Açık" if j in open_depots else "Açılmadı",
    })
hospital_table = pd.DataFrame(
    hospital_rows
)
display(
    hospital_table
)

# =====================================================================
# AÇILAN DEPOLAR
# =====================================================================
print(
    "\nAçılan depolar:"
)
for j in open_depots:
    load = sum(
        DEMAND_POINTS[i]["demand"]
        for i in DEMAND_POINTS
        if assignments[i] == j
    )
    depot = DEPOT_CANDIDATES[j]
    print(
        f"- {depot['name']}: "
        f"{load:,} paket ihtiyaç / "
        f"{depot['capacity']:,} kapasite | "
        f"{depot['ground_area_m2']:,} m² | "
        f"{depot['cabinet_count']} dolap | "
        f"{depot['energy_kwh_day']:.2f} kWh/gün"
    )

# =====================================================================
# TOPLAM ENERJİ
# =====================================================================
total_energy = sum(
    DEPOT_CANDIDATES[j]["energy_kwh_day"]
    for j in open_depots
)
print(
    f"\nSeçilen depo sayısı: {p_star}"
)
print(
    f"Minimax değeri: "
    f"{z_star:,.2f} ihtiyaç-km"
)
print(
    f"Toplam enerji: "
    f"{total_energy:.2f} kWh/gün"
)
print(
    f"Toplam enerji sınırı: "
    f"{MAX_TOTAL_ENERGY_KWH_DAY:.2f} kWh/gün"
)

# =====================================================================
# AÇIK DEPOLARIN TALEP NOKTALARINA ULAŞIM SÜRELERİ
# =====================================================================
# Modelin açtığı depoların her talep noktasına ulaşım süresini tablo halinde gösterir.
travel_time_rows = []
for j in open_depots:
    depot = DEPOT_CANDIDATES[j]
    for i, demand in DEMAND_POINTS.items():
        travel_time_rows.append({
            "depo_kodu": j,
            "depo": depot["name"],
            "talep_noktası": demand["name"],
            "mesafe_km": distance[(i, j)],
            "t_saat": travel_time[(i, j)],
            "t_dakika": travel_time[(i, j)] * 60,
            "bozulma_kısıtı": (
                "Uygun"
                if travel_time[(i, j)] <= MAX_TRAVEL_HOURS
                else "Uygun değil"
            )
        })
travel_time_table = pd.DataFrame(travel_time_rows).sort_values(
    by=["depo_kodu", "t_saat"],
    ascending=[True, True]
).reset_index(drop=True)
display(travel_time_table)

# =====================================================================
# GÖRSELLEŞTİRME
# =====================================================================
show_solution(
    distance,
    open_depots,
    assignments,
    (
        "İzmir İlaç Depolama Ağı (Yalnız Toplam Enerji Sınırı): "
        f"Açılan Depolar ve İlçe Atamaları ({p_star} depo)"
    ),
)
```
